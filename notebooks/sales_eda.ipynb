```python
# Step 1: Import libraries
import pandas as pd
import numpy as np

```

```python

# Step 2: Load the data
file_path = r"C:\Users\aishw\OneDrive\Desktop\Analyst Projects\Sales-performance\data\superstore_sales.csv"
  # Adjust path if needed
df = pd.read_csv(file_path, encoding='latin1')  # 'latin1' handles special characters


```

```python
# Step 3: Take a peek at the data
df.head()
```

```python
# Check dataset shape and column info
print("Shape of dataset:", df.shape)
print("\nColumn Info:\n")
df.info()

# Summary statistics
df.describe(include='all').T.head(15)
```

```python
# Clean column names: remove spaces, make lowercase
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
df.head(2)
```

```python
# Check for missing values
df.isna().sum()

# Check for duplicates
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
# --- Overall KPIs ---
total_sales = df['sales'].sum()
total_profit = df['profit'].sum()
total_orders = df['order_id'].nunique()
avg_discount = df['discount'].mean()

print(f"Total Sales: ${total_sales:,.0f}")
print(f"Total Profit: ${total_profit:,.0f}")
print(f"Total Orders: {total_orders:,}")
print(f"Average Discount: {avg_discount:.2%}")
```

```python
# --- Sales by Category ---
sales_by_category = df.groupby('category')['sales'].sum().sort_values(ascending=False)
print("\nSales by Category:\n", sales_by_category)

# --- Sales by Region ---
sales_by_region = df.groupby('region')['sales'].sum().sort_values(ascending=False)
print("\nSales by Region:\n", sales_by_region)
```

```python
df['order_date'].head()
print(df['order_date'].dtype)
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')
```

```python
df['order_date'].isna().sum()
```

```python
monthly_sales = df.groupby(df['order_date'].dt.to_period('M'))['sales'].sum().reset_index()
monthly_sales['order_date'] = monthly_sales['order_date'].astype(str)

import plotly.express as px
fig = px.line(monthly_sales, x='order_date', y='sales', title='Monthly Sales Trend')
fig.show()

```

```python
import plotly.express as px
cat_sales_profit = df.groupby('category')[['sales','profit']].sum().reset_index()
fig = px.bar(cat_sales_profit, x='category', y=['sales','profit'], barmode='group',
             title='Sales and Profit by Category', text_auto=True)
fig.show()
```

```python
region_sales_profit = df.groupby('region')[['sales','profit']].sum().reset_index()
fig = px.bar(region_sales_profit, x='region', y=['sales','profit'], barmode='group',
             title='Sales and Profit by Region', text_auto=True)
fig.show()

```

```python
top_subcats = df.groupby('sub-category')['sales'].sum().sort_values(ascending=False).head(10).reset_index()
fig = px.bar(top_subcats, x='sub-category', y='sales', color='sales',
             title='Top 10 Sub-Categories by Sales', text='sales')
fig.show()
```
